```python
import spacy
from nltk import ngrams
nlp=spacy.load("en_core_web_sm")
doc=nlp("AI is simulation of human intelligence by machine. IT includes expects system nlp speech recognition and machine vision")
bigrams=ngrams(doc,2)
print(list(bigrams))
```
